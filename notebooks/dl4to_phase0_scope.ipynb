```python
import pandas as pd

df = pd.read_csv("../data/clean/meltpool_master_with_VED.csv")

print("Dataset shape:", df.shape)
print("\nMaterials present:", df["material"].unique())


# Global validity window

P_min, P_max = df["P_W"].min(), df["P_W"].max()
v_min, v_max = df["v_mm_per_s"].min(), df["v_mm_per_s"].max()

print("\n--- GLOBAL SURROGATE VALIDITY WINDOW ---")
print(f"Laser Power P ∈ [{P_min:.1f}, {P_max:.1f}] W")
print(f"Scan Speed v ∈ [{v_min:.1f}, {v_max:.1f}] mm/s")


# Material-wise validity

print("\n--- MATERIAL-SPECIFIC WINDOWS ---")
for mat in sorted(df["material"].unique()):
    sub = df[df["material"] == mat]
    print(f"\nMaterial: {mat}")
    print(f"  P ∈ [{sub['P_W'].min():.1f}, {sub['P_W'].max():.1f}] W")
    print(f"  v ∈ [{sub['v_mm_per_s'].min():.1f}, {sub['v_mm_per_s'].max():.1f}] mm/s")


# Fixed assumptions check

print("\n--- FIXED PROCESS ASSUMPTIONS (FROM DATA) ---")
print("Unique hatch spacings (mm):", df["h_mm"].unique())
print("Unique layer thicknesses (mm):", df["t_mm"].unique())


# Energy metrics sanity

print("\n--- ENERGY METRICS RANGE ---")
print(f"LED ∈ [{df['LED_J_per_mm'].min():.3f}, {df['LED_J_per_mm'].max():.3f}] J/mm")
print(f"VED ∈ [{df['VED_J_per_mm3'].min():.3f}, {df['VED_J_per_mm3'].max():.3f}] J/mm³")
```
